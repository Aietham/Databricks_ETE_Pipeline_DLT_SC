# Dynamic Capabilities

```python
dbutils.widgets.text('filename','')
```

```python
filename = dbutils.widgets.get('filename')
```

# Reading Data

```python
df = spark.readStream.format("cloudFiles") \
  .option("cloudFiles.format", "parquet") \
  .option("cloudFiles.schemaLocation", f"/Volumes/project/_ops/autoloader/_schema/{filename}") \
  .option("cloudFiles.schemaEvolutionMode", "addNewColumns") \
  .option("cloudFiles.maxFilesPerTrigger", 10) \
  .load(f"/Volumes/project/default/src_volume/{filename}")
```

```python
f"/Volumes/project/_ops/autoloader/_schema/{filename}"
```

```python
f'/Volumes/project/source_volumes/src_{filename}'
```

# Writing Data

```python
query = (
  df.writeStream
    .option("checkpointLocation", f"/Volumes/project/_ops/autoloader/_checkpoints/{filename}")
    .trigger(availableNow=True) 
    .table(f"project.bronze.{filename}")
)

query.awaitTermination()
```
