# Creating Initial Base Objects

### External Location For Managed Catalog


```python
storage_account = "lakehouseadlsdev"
catalog_conatiner = "catalog-managed"
catalog_name = "retail-dev"
source_container = "source"
```

```python
spark.sql(f"""CREATE EXTERNAL LOCATION IF NOT EXISTS ext_catalog_location
  URL 'abfss://{catalog_conatiner}@{storage_account}.dfs.core.windows.net/catalog'
  WITH (STORAGE CREDENTIAL adls_connector)""")
```

### Creating Managed Catalog

```python
spark.sql(f"""CREATE CATALOG IF NOT EXISTS project
  MANAGED LOCATION  'abfss://{catalog_conatiner}@{storage_account}.dfs.core.windows.net/catalog'""")
```

### External Location for the Source

```python
spark.sql(f"""CREATE EXTERNAL LOCATION IF NOT EXISTS ext_source_location
  URL 'abfss://{source_container}@{storage_account}.dfs.core.windows.net'
  WITH (STORAGE CREDENTIAL adls_connector)""")
```
